```python
from langchain_core.tools import tool
```

```python
# step-1 create a function

def multiply(a,b):
    """multiply two number"""
    
    return a * b 
```

```python
#step-2 add hints type 

def multiply(a:int , b:int)->int:
    """multiply two number"""
    return a * b 
    
```

```python
#step-3 tool decoretor 
@tool 
def multiply(a:int , b:int)->int:
    """multiply two number"""
    
    return a * b 
```

```python
result = multiply.invoke({"a":3, "b":4})
print("your result is :",result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

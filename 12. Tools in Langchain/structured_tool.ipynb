```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
%pip install -U langchain langchain-core langsmith langchain-community langchain-experimental
```

```python
class MultiplyInputModel(BaseModel):
    
    a:int = Field(required =True , description = "The first number to add")
    b:int = Field(required =True, description = "The second number to add")
    
    
```

```python
def multiply_func(a:int ,b:int) ->int:
    return a * b 
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_func,
    name = 'multiply',
    description = "Multiply two number",
    args_schema = MultiplyInputModel
)
```

```python
result = multiply_tool.invoke({'a':3, 'b':4})
print(result)
print(multiply_tool.name)
```

```python

import random
class FakeLLM:

  def __init__(self):
    print('LLM Created')

  def predict(self,prompt):
    respone_list = [
        'Dhaka is the capital city of bangladesh',
        'BPL is the worst cricket leguage ever',
        'AI stands for Artificial Intelligence'
    ]

    return {'respone' : random.choice(respone_list)}


```

```python
llm = FakeLLM()
```

```python
llm.predict("What is the capital of bangladesh?")
```

```python
class FakePromptTemplate:

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def format(self, input_dct):
    return self.template.format(**input_dct)
```

```python
template = FakePromptTemplate(
    template= 'Write a story about the {topic}',
    input_variables= ['topic']
)
```

```python
prompt = template.format({'topic' : 'cricket'})
```

```python
llm.predict(prompt)
```

```python
class FakeLLMChain:

  def __init__(self,llm,prompt):
    self.llm = llm
    self.prompt = prompt

  def run (self,input_dict):
    final_prompt = self.prompt.format(input_dict)
    result = self.llm.predict(final_prompt)
```

```python
template = FakePromptTemplate(
    template= 'Write a story about the {topic}',
    input_variables= ['topic']
)
```

```python
llm1 = FakeLLM()
```

```python
chain = FakeLLMChain(llm1,template)
```

```python
chain.run({'topic' : 'cricket'})
```
